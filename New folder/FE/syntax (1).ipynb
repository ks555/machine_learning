### LabelEncoder

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('loan_data.csv')
df.head()
```

```python
df.info()
```

```python
X = df.drop(columns='loan_status')
y = df['loan_status']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(
    X,
    y,
    train_size=0.8,
    random_state=42
)
```

```python
X
```

```python
cat_cols = X.select_dtypes(include='object')
cat_cols.nunique()
```

```python
cat_cols.head()
```

```python
xtrain
```

```python
'''.
label_encoder = LabelEncoder()
encoded_values_train = label_encoder.fit_transform(xtrain) 
Here xtrain contains 13 columns which includes both numerical and 
categorical columns but fit_transform accepts only one categorical column.
# ValueError: y should be a 1d array, got an array of shape (36000, 13) instead.
'''
#correct syntax
label_encoder = LabelEncoder()
encoded_values_train = label_encoder.fit_transform(xtrain['person_gender'])
encoded_values_train 
```

```python
encoded_values_test = label_encoder.transform(xtest['person_gender'])
# if we pass two or more categorical columns => ValueError
encoded_values_test 
```

### OneHotEncoder

```python
xtrain
```

```python
cat_cols = X.select_dtypes(include='object').columns
```

```python
'''
from sklearn.preprocessing import OneHotEncoder
onehot_encoder = OneHotEncoder(
    sparse_output=True, or False
    handle_unknown='error' or 'ignore'
)
encoded_values=onehot_encoder.fit_transform(xtrain[cat_cols])
fit_transform()
 -Accepts 2D
 -fit() -> learns the pattern
 1) selects the unique values from all the columns
 2) sorts by Ascending order
 3) creates new columns
-transform() -> encodes the values,Assigns 1 if the label is present in a row else 0
sparse_output=True,
returns => compressed sparse matrix.
sparse_output=False,
returns => encoded values.

encoded_values=onehot_encoder.transform(xtest[cat_cols])
-Accepts 2D
-returns the values based on sparse_output
-handle_unknown='error'
for unknown labels it throws error(ValueError)
-handle_unkown ='ignore'
Assigns Zero for unknown labels
'''
from sklearn.preprocessing import OneHotEncoder
onehot_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown='ignore'
)
encoded_values=onehot_encoder.fit_transform(xtrain[cat_cols])
```

```python
encoded_cols = onehot_encoder.get_feature_names_out()
```
